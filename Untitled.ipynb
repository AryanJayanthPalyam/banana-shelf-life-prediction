```python
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from PIL import Image
import torch.nn as nn

# Map your exact variety names from CSV to integer values
VARIETY_MAP = {'yellaki': 0, 'robusta': 1, 'Redbanana': 2}

image_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
```

```python
class BananaDataset(Dataset):
    def __init__(self, csv_path, root_dir, transform=image_transform):
        self.df = pd.read_csv(csv_path)
        self.root_dir = root_dir
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(f"{self.root_dir}/{row['img_path']}").convert("RGB")
        img = self.transform(img)
        variety_idx = VARIETY_MAP[row['variety']]
        label = torch.tensor(row['days_until_death'], dtype=torch.float32)
        return img, torch.tensor(variety_idx, dtype=torch.long), label
```

```python
class BananaCNNMulti(nn.Module):
    def __init__(self, n_varieties=3, embed_dim=8):
        super().__init__()
        self.resnet = models.resnet18(weights='IMAGENET1K_V1')
        self.resnet.fc = nn.Identity()
        self.variety_embed = nn.Embedding(n_varieties, embed_dim)
        self.fc = nn.Sequential(
            nn.Linear(512 + embed_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )
    def forward(self, x_img, x_variety):
        img_feat = self.resnet(x_img)
        var_feat = self.variety_embed(x_variety)
        x = torch.cat([img_feat, var_feat], dim=1)
        return self.fc(x).squeeze(-1)
```

```python
csv_path = "Banana/labels.csv"
root_dir = "Banana"
dataset = BananaDataset(csv_path, root_dir)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)

model = BananaCNNMulti(n_varieties=3)
criterion = nn.L1Loss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

epochs = 10
for epoch in range(epochs):
    model.train()
    running_loss = 0
    for imgs, varieties, labels in dataloader:
        preds = model(imgs, varieties)
        loss = criterion(preds, labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
    print(f"Epoch {epoch+1}: Loss={running_loss / len(dataset):.3f}")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

model.eval()
all_preds, all_labels = [], []
with torch.no_grad():
    for imgs, varieties, labels in dataloader:
        preds = model(imgs, varieties)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

mae = mean_absolute_error(all_labels, all_preds)
rmse = mean_squared_error(all_labels, all_preds)
r2 = r2_score(all_labels, all_preds)
print(f"MAE: {mae:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R2: {r2:.3f}")
```
